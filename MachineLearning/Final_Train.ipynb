```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Train2.csv')
```

```python
df.head(5)
```

```python
df.shape
```

```python
df1 = pd.read_csv('Train.csv')
```

```python
df1['url'].dtype
```

```python
df1 = df1.drop(['Unnamed: 0'], axis = 1)
```

```python
df1['url'] = 'http://' + df1['url'].astype(str)
df1
```

```python
phish = df1[df1['Phish?'] == 1].sample(56000)
```

```python
phish
```

```python
leg = df1[~df1['url'].isin(df['url'])]
```

```python
leg = leg[leg['Phish?'] == 0]
```

```python
leg = leg.sample(54000)
```

```python
leg
```

```python
merge = pd.concat([phish, leg], axis = 0)
```

```python
merge = merge.sample(frac = 1).reset_index(drop = True)
```

```python
merge
```

```python
final = pd.concat([df, merge], axis = 0)
```

```python
final
```

```python
final = final.sample(frac = 1).reset_index(drop = True)
```

```python
final.shape
```

```python
final.to_csv('Final_Train.csv')
```
